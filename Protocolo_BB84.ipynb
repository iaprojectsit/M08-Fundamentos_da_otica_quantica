```python
import numpy as np

def simular_bb84(tamanho_chave=50):
    """
    Simulação do Protocolo Quântico de Distribuição de Chaves (BB84)
    explorando a polarização da luz.
    0 -> Base Retilínea (+), 1 -> Base Diagonal (X)
    """
    np.random.seed(101)
    # 1. Alice gera os bits e escolhe suas bases aleatoriamente
    bits_alice = np.random.randint(2, size=tamanho_chave)
    bases_alice = np.random.randint(2, size=tamanho_chave)

    # 2. Bob escolhe aleatoriamente suas bases de medição
    bases_bob = np.random.randint(2, size=tamanho_chave)

    # 3. Bob mede o feixe laser
    bits_bob = np.zeros(tamanho_chave, dtype=int)
    for i in range(tamanho_chave):
        if bases_alice[i] == bases_bob[i]:
            # Bases iguais = medição perfeita (probabilidade 100% no caso ideal)
            bits_bob[i] = bits_alice[i]
        else:
            # Bases diferentes = resultado puramente aleatório (50% probabilidade)
            bits_bob[i] = np.random.randint(2)

    # 4. Etapa Pública: Reconciliação (Sifting)
    # Alice e Bob revelam apenas as bases utilizadas, descartando medições com bases diferentes
    bases_compativeis = (bases_alice == bases_bob)
    chave_segura_alice = bits_alice[bases_compativeis]
    chave_segura_bob = bits_bob[bases_compativeis]

    # 5. Cálculo do Quantum Bit Error Rate (QBER)
    erros = np.sum(chave_segura_alice != chave_segura_bob)
    qber = (erros / len(chave_segura_alice)) * 100 if len(chave_segura_alice) > 0 else 0

    print("--- SIMULAÇÃO DE DISTRIBUIÇÃO DE CHAVES (BB84) ---")
    print(f"Bits Gerados por Alice: {bits_alice[:15]}...")
    print(f"Bases Usadas por Alice: {bases_alice[:15]}...")
    print(f"Bases Usadas por Bob:   {bases_bob[:15]}...")
    print(f"Chave Segura (Alice):   {chave_segura_alice}")
    print(f"Chave Segura (Bob):     {chave_segura_bob}")
    print(f"\nTamanho da Chave Retida: {len(chave_segura_alice)} bits (~50% do original)")
    print(f"Taxa de Erro (QBER): {qber:.2f}%")

simular_bb84()
```
